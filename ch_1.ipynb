# **Part A: Indexing and Slicing**

```python
print('='*50)
print('creating the matrix')
print('='*50)

matrix = np.arange(0, 20,1).reshape((5, 4))
```

```python
# Display shape
print('='*50)
print('shape of the matrix ')
print('='*50)
print(matrix.shape)
print(f'number of rows    : {matrix.shape[0]}')
print(f'number of columns : {matrix.shape[1]}')
```

```python
# number dimensions
print('='*50)
print('number of dimensions ')
print('='*50)
print(f'number of dimensions : {matrix.ndim}')
```

```python
# data type
print('='*50)
print('matrix data type')
print('='*50)
print(f'data type : {matrix.dtype}')
```

```python
# total number of elements
print('='*50)
print('total number of elements')
print('='*50)
number_of_elemets = matrix.shape[0] * matrix.shape[1]
print(f'total : {number_of_elemets}')
```

```python
# Access a single element (row at index 3, column at index 2)

print('='*50)
print('Access a single element (row at index 3, column at index 2)')
print('='*50)
print(f'value : {matrix[3:4, 2:3][0][0]}')
```

```python
# access the last element using negative indices
print('='*50)
print('access the last element using negative indices')
print('='*50)
print(f'value : {matrix[-1, -1]}')
```

```python
# Extract the first row

print('='*50)
print('Extract the first row')
print('='*50)
print(matrix[:1][0])
```

```python
# the last column

print('='*50)
print('Extract the last column')
print('='*50)
print(matrix[:, -1:])
```

```python
# entire second column

print('='*50)
print('entire second column')
print('='*50)
print(matrix[:, 1:2])
```

```python
# Use slicing to extract the central block (rows at indices 1 to 3, columns at indices 1 to 2).

print('='*50)
print('rows at indices 1 to 3, columns at indices 1 to 2')
print('='*50)
print(matrix[1:4, 1:3])
```

```python
# Extract every other row, then reverse the order of the rows.

print('='*50)
print('reverse the order of the rows')
print('='*50)
print(np.flip(matrix, 1))
```

```python
# Select rows 0, 2, and 4 using a list of indices. What is the difference compared to slicing?

print('='*50)
print('Select rows 0, 2, and 4')
print('='*50)
print(matrix[[0, 2, 4]])
```

```python
# Using a Boolean mask, select even values greater than 8, then replace them with 0.

print('='*50)
print('select even values greater than 8, then replace them with 0')
print('='*50)
boolean_mask= (matrix % 2 == 0) & (matrix > 8) 
matrix[boolean_mask] = 0
print(matrix)
```

```python
# Find the positions (indices) of values divisible by 5.
print('='*50)
print('find indeces of values divisible by 5')
print('='*50)
print(np.argwhere(matrix % 5 == 0))
```

```python
# make a copy

mcopy = matrix.copy()

print('='*50)
print('copy of the matrix before editing')
print('='*50)
print(mcopy)
# edit the value of index [0:1, 0:1]

mcopy[0:1, 0:1] = 2


```

```python
# edited copy of the matrix 

print('='*50)
print('edited copy of the matrix')
print('='*50)
mcopy
```
