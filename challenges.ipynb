# **Part A: Indexing and Slicing**

```python
print_title('creating the matrix')

matrix = np.arange(0, 20,1).reshape((5, 4))
```

```python
# Display shape
print_title('shape of the matrix ')
print(matrix.shape)
print(f'number of rows    : {matrix.shape[0]}')
print(f'number of columns : {matrix.shape[1]}')
```

```python
# number dimensions
print_title('number of dimensions ')
print(f'number of dimensions : {matrix.ndim}')
```

```python
# data type
print_title('matrix data type')
print(f'data type : {matrix.dtype}')
```

```python
# total number of elements
print_title('total number of elements')
number_of_elemets = matrix.shape[0] * matrix.shape[1]
print(f'total : {number_of_elemets}')
```

```python
# Access a single element (row at index 3, column at index 2)

print_title('Access a single element (row at index 3, column at index 2)')
print(f'value : {matrix[3:4, 2:3][0][0]}')
```

```python
# access the last element using negative indices

print_title('access the last element using negative indices')
print(f'value : {matrix[-1, -1]}')
```

```python
# Extract the first row

print_title('Extract the first row')
print(matrix[:1][0])
```

```python
# the last column

print_title('Extract the last column')
print(matrix[:, -1:])
```

```python
# entire second column

print_title('entire second column')
print(matrix[:, 1:2])
```

```python
# Use slicing to extract the central block (rows at indices 1 to 3, columns at indices 1 to 2).

print_title('rows at indices 1 to 3, columns at indices 1 to 2')
print(matrix[1:4, 1:3])
```

```python
# Extract every other row, then reverse the order of the rows.

print_title('reverse the order of the rows')
print(np.flip(matrix, 1))
```

```python
# Select rows 0, 2, and 4 using a list of indices. What is the difference compared to slicing?

print_title('Select rows 0, 2, and 4')
print(matrix[[0, 2, 4]])
```

```python
# Using a Boolean mask, select even values greater than 8, then replace them with 0.

print_title('select even values greater than 8, then replace them with 0')
boolean_mask= (matrix % 2 == 0) & (matrix > 8) 
matrix[boolean_mask] = 0
print(matrix)
```

```python
# Find the positions (indices) of values divisible by 5.
print_title('find indeces of values divisible by 5')
print(np.argwhere(matrix % 5 == 0))
```

```python
# make a copy

mcopy = matrix.copy()

print_title('copy of the matrix before editing')
print(mcopy)
# edit the value of index [0:1, 0:1]

mcopy[0:1, 0:1] = 2


```

```python
# edited copy of the matrix 

print_title('edited copy of the matrix')
mcopy
```

```python
# original matrix

print_title('original matrix')
matrix
```

# **Part B: NumPy with the Titanic dataset**

```python
print_title('extract age and fare as NumPy arrays.')
np_age= df['age'].to_numpy()
np_fare= df['fare'].to_numpy()
```

```python
print_title('Count the missing values in age.')
print(f' missing values in "age" : {np.isnan(np_age).sum()}')
```

```python
print_title('Compare the regular mean with the mean that ignores NaN values.')

print(f' regulare mean               : {np.mean(np_age)}')
print(f' mean that ignore NaN values : {np.nanmean(np_age)}')
```

```python
print_title('Age statistics')
print(f'mean               : {np.nanmean(np_age)}')
print(f'median             : {np.nanmedian(np_age)}')
print(f'standard deviation : {np.nanstd(np_age)}')
print(f'minimum            : {np.nanmin(np_age)}')
print(f'maximum            : {np.nanmax(np_age)}')

print_title('fare statistics')
print(f'mean               : {np.nanmean(np_fare)}')
print(f'median             : {np.nanmedian(np_fare)}')
print(f'standard deviation : {np.nanstd(np_fare)}')
print(f'minimum            : {np.nanmin(np_fare)}')
print(f'maximum            : {np.nanmax(np_fare)}')

age_q1 = np.nanpercentile(np_age, 25)
age_q3 = np.nanpercentile(np_age, 75)

print_title('25 and 75 percentiles')
print(f'25th percentile    : {age_q1}')
print(f'75th percentile    : {age_q3}')
```

```python
print_title('Calculate the IQR (Interquartile Range) of fare')
fare_q1 = np.nanpercentile(np_fare, 25)
fare_q3 = np.nanpercentile(np_fare, 75)
fare_iqr = fare_q3 - fare_q1
print(f'IQR            :{fare_iqr}')

print_title('calculate the upper bound using Q3 + 1.5 x IQR.')
upper_bound = fare_q3 + 1.5 * fare_iqr
print(f'upper bound    : {upper_bound}')

print_title('count the outliers using a Boolean mask.')
print(f'outliers count : {(np_fare > upper_bound).sum()}')
```

```python
print_title('Apply np.log1p() to fare')
log_fare = np.log1p(np_fare)

print_title('the distributions before log1p transformation (mean and median).')
print(f'mean               : {np.nanmean(np_fare)}')
print(f'median             : {np.nanmedian(np_fare)}')

print_title('the distributions after log1p transformation (mean and median).')
print(f'log1p mean         : {np.nanmean(log_fare)}')
print(f'log1p median       : {np.nanmedian(log_fare)}')
```

```python
print_title('Create an array classifying passengers as either child or adult (using 12 years as the age threshold)')

passangers_classification = np.select([np.isnan(np_age),
                                       np_age <= 12,
                                       np_age > 12],
                                       ['unknown',
                                       'child',
                                       'adult'],
                                       default='unknown')

print(f'first 20 values    : {passangers_classification[:20]}...')


print_title('count the number of children.')
print(f'number of children : {(passangers_classification == 'child').sum()}')
```

```python
print_title('Calculate the correlation between age and fare')
mask = ~np.isnan(np_age) & ~np.isnan(np_fare)
clean_fare= np_fare[mask]
clean_age= np_age[mask]
corr = np.corrcoef(clean_age, clean_fare)[0,1]
print(corr)
```

# **Matplotlib**

```python

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

bins_list = [10, 20, 40]

for ax, bins in zip(axes, bins_list):
    ax.hist(df['age'].dropna(), bins=bins, edgecolor='black')
    ax.set_title(f'Age Distribution ({bins} bins)')
    ax.set_xlabel('Age')
    ax.set_ylabel('Number of Passengers')

plt.tight_layout()
plt.show()
```

```python
class_counts = df['pclass'].value_counts().sort_index()

plt.figure(figsize=(7, 4))

plt.bar(class_counts.index, class_counts.values, edgecolor='black')

plt.title('Number of Passengers per Class')
plt.xlabel('Passenger Class')
plt.ylabel('Number of Passengers')
plt.xticks([1, 2, 3])

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df['age'], df['fare'], alpha=0.5)

plt.title('Age vs Fare')
plt.xlabel('Age')
plt.ylabel('Fare')

plt.show()
```

```python
class_counts = df['pclass'].value_counts().sort_index()
survival_by_sex = df.groupby('sex')['survived'].mean()

fig, axes = plt.subplots(2, 2, figsize=(12, 9))

# 1. Age histogram
axes[0, 0].hist(df['age'].dropna(), bins=20, edgecolor='black')
axes[0, 0].set_title('Age Distribution')
axes[0, 0].set_xlabel('Age')
axes[0, 0].set_ylabel('Count')

# 2. Passengers per class

axes[0, 1].bar(class_counts.index, class_counts.values, edgecolor='black')
axes[0, 1].set_title('Passengers per Class')
axes[0, 1].set_xlabel('Class')
axes[0, 1].set_ylabel('Count')
axes[0, 1].set_xticks([1, 2, 3])

# 3. Age vs Fare
axes[1, 0].scatter(df['age'], df['fare'], alpha=0.5)
axes[1, 0].set_title('Age vs Fare')
axes[1, 0].set_xlabel('Age')
axes[1, 0].set_ylabel('Fare')

# 4. Survival by sex

axes[1, 1].bar(survival_by_sex.index, survival_by_sex.values)
axes[1, 1].set_title('Survival Rate by Sex')
axes[1, 1].set_xlabel('Sex')
axes[1, 1].set_ylabel('Survival Rate')

fig.suptitle('Titanic Dataset — Data Visualization')

plt.tight_layout()
plt.show()
```
